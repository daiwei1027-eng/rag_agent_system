```python
from datasets import load_dataset

dataset = load_dataset(
    "roneneldan/TinyStories",
    split="train",
)

print(dataset[0]["text"])
```

```python
from transformers import AutoTokenizer
from rag_agent_system.ingestion import build_story_chunks, chunk_story

chunk_tokenizer = AutoTokenizer.from_pretrained(
    "Qwen/Qwen2.5-0.5B-Instruct",
    use_fast=True,
)

embedding_tokenizer = AutoTokenizer.from_pretrained(
    "sentence-transformers/all-MiniLM-L6-v2",
    use_fast=True,
)

# 先验证 100 篇，正式语料范围随后固定。
all_chunks = build_story_chunks(
    (dataset[i]["text"] for i in range(100)),
    chunk_tokenizer,
    embedding_tokenizer=embedding_tokenizer,
    max_tokens=128,
    overlap=32,
)

print("Chunk 数：", len(all_chunks))
print(all_chunks[0])
```
